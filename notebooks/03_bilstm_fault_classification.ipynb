```python
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt
import random

np.random.seed(42)
tf.random.set_seed(42)
random.seed(42)
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
import h5py
path = "/content/drive/My Drive/Datasets/TrainAndTestDataset.mat"

file =h5py.File(path,'r')
```

```python
X = np.array(file['Xall']).transpose(2,1,0)
y = np.array(file['Yall']).transpose()
labels = np.argmax(y, axis=1)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_temp, y_train, y_temp, y_train_lbl, y_temp_lbl = train_test_split(
    X, y, labels,
    test_size=0.3,
    random_state=42,
    stratify=labels
)

X_val, X_test, y_val, y_test, y_val_lbl, y_test_lbl = train_test_split(
    X_temp, y_temp, y_temp_lbl,
    test_size=0.50,
    stratify=y_temp_lbl,
    random_state=42
)
```

```python
model = models.Sequential([

    layers.Input(shape=(200,6)),

    # First BiLSTM Layer (captures detailed sequence)
    layers.Bidirectional(
        layers.LSTM(128, return_sequences=True)
    ),
    layers.BatchNormalization(),

    # Second BiLSTM Layer
    layers.Bidirectional(
        layers.LSTM(64)
    ),

    # Dense Feature Learning
    layers.Dense(128, activation='relu'),
    layers.BatchNormalization(),
    layers.Dropout(0.4),

    layers.Dense(64, activation='relu'),
    layers.Dropout(0.3),

    layers.Dense(5, activation='softmax')
])
```

```python
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

model.summary()
```

```python
from tensorflow.keras.utils import plot_model

plot_model(
    model,
    to_file='bilstm_architecture.png',
    show_shapes=True,
    show_layer_names=True,
    expand_nested=True,
    dpi=300
)
```

```python
early_stop = EarlyStopping(
    monitor='val_loss',
    patience=7,
    restore_best_weights=True
)

reduce_lr = ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.5,
    patience=3,
    min_lr=0.00001
)
```

```python
history = model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=1000,
    batch_size=64,
    callbacks=[early_stop, reduce_lr],
    verbose=1
)
```

```python
test_loss, test_acc = model.evaluate(X_test, y_test)
```

```python
#Training curve
plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.plot(history.history['accuracy'], label='Train')
plt.plot(history.history['val_accuracy'], label='Val')
plt.title("Accuracy")
plt.legend()

plt.subplot(1,2,2)
plt.plot(history.history['loss'], label='Train')
plt.plot(history.history['val_loss'], label='Val')
plt.title("Loss")
plt.legend()

plt.show()
```

```python
#Confusion Matrix
y_pred = model.predict(X_test)
y_pred_cls = np.argmax(y_pred, axis=1)
y_true = np.argmax(y_test, axis=1)

cm = confusion_matrix(y_true, y_pred_cls)

plt.figure(figsize=(6,5))
sns.heatmap(cm, annot=True, fmt='d',
            xticklabels=["LG","LL","LLG","LLL","None"],
            yticklabels=["LG","LL","LLG","LLL","None"])
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("BiLSTM Confusion Matrix")
plt.show()
```

```python
#Classification Report
print(classification_report(
    y_true,
    y_pred_cls,
    target_names=["LG","LL","LLG","LLL","None"]
))
```
